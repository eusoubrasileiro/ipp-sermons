```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Load the multilingual model
model = SentenceTransformer("sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")

# Subtopics list
subtopics = [
    'Pensamento Cristão', 'Educação', 'Vida da Mente', 'Certeza da Salvação', 'Consciência',
    'Aconselhamento', 'Coragem & Ousadia', 'Morte & Morrer', 'Depressão', 'Dieta & Exercício',
    # Add the rest of the subtopics here...
]

# Example sermons (replace with your actual sermons)
sermons = [
    "Como desenvolver coragem e ousadia na vida cristã.",
    "Entendendo a certeza da salvação através das Escrituras.",
    "A importância da oração e da vida devocional."
]

# Encode subtopics and sermons
subtopics_embeddings = model.encode(subtopics, convert_to_tensor=True)
sermon_embeddings = model.encode(sermons, convert_to_tensor=True)

# Define cutoff and max subtopics per sermon
SIMILARITY_CUTOFF = 0.5
MAX_SUBTOPICS = 3

# Classify sermons into subtopics
for i, sermon_embedding in enumerate(sermon_embeddings):
    similarities = cosine_similarity(
        sermon_embedding.cpu().numpy().reshape(1, -1), 
        subtopics_embeddings.cpu().numpy()
    ).flatten()
    
    # Get subtopics above cutoff
    relevant_indices = np.where(similarities >= SIMILARITY_CUTOFF)[0]
    sorted_indices = relevant_indices[np.argsort(-similarities[relevant_indices])]
    top_indices = sorted_indices[:MAX_SUBTOPICS]
    
    # Map indices to subtopics
    classified_subtopics = [subtopics[idx] for idx in top_indices]
    
    # Print results
    print(f"Sermon {i+1}: {sermons[i]}")
    print(f"Classified Subtopics: {classified_subtopics}")
    print("---------")
```
